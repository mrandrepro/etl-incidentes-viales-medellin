## 1. Importar librerías y cargar datos

```python
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go

# Cargar dataset
df = pd.read_csv(
    '../1_raw_data/incidentes_viales.csv',
    encoding='utf-8',
    on_bad_lines='skip',
    low_memory=False
)

print(f'Filas: {df.shape[0]:,}')
print(f'Columnas: {df.shape[1]}')
df.head()
```

## 2. Vista general del dataset

```python
df.info()
```

```python
df.describe(include='all')
```

## 3. Análisis de valores nulos

```python
nulos = df.isnull().sum().reset_index()
nulos.columns = ['columna', 'nulos']
nulos['porcentaje'] = (nulos['nulos'] / len(df) * 100).round(2)
nulos = nulos[nulos['nulos'] > 0].sort_values('nulos', ascending=False)
display(nulos)

fig = px.bar(
    nulos,
    x='porcentaje',
    y='columna',
    orientation='h',
    title='Porcentaje de valores nulos por columna',
    labels={'porcentaje': '% de nulos', 'columna': 'Columna'},
    color='porcentaje',
    color_continuous_scale='Reds',
    text='porcentaje'
)
fig.update_traces(texttemplate='%{text:.1f}%', textposition='outside')
fig.update_layout(yaxis={'categoryorder': 'total ascending'}, showlegend=False)
fig.show()

print('>>> PROBLEMA: BARRIO (7.0%), CBML (6.7%) y COMUNA (4.7%) tienen nulos significativos.')
print('>>> Decisión ETL: imputar con DESCONOCIDO para mantener integridad referencial.')
```

## 4. Detección de duplicados

```python
dup_filas = df.duplicated().sum()
dup_expediente = df['EXPEDIENTE'].duplicated().sum()

print(f'Filas completamente duplicadas: {dup_filas}')
print(f'EXPEDIENTE duplicado: {dup_expediente}')
print('>>> Decisión ETL: eliminar las 7 filas duplicadas exactas con drop_duplicates().')
print('>>> Los 612 EXPEDIENTE duplicados se conservan (pueden ser múltiples involucrados).')
```

## 5. Validación de tipos de datos

```python
print('Valores únicos de AÑO antes de limpiar:')
print(sorted(df['AÑO'].astype(str).unique()))
print('>>> PROBLEMA: AÑO tiene tipo object con valores como 2019\\r (salto de línea embebido).')
print('>>> Decisión ETL: strip + replace \\r, luego castear a int.')

print('\nEjemplos FECHA_ACCIDENTE:', df['FECHA_ACCIDENTE'].head(3).tolist())
print('Ejemplos FECHA_ACCIDENTES:', df['FECHA_ACCIDENTES'].head(3).tolist())
print('>>> NOTA: Columna FECHA_ACCIDENTES es redundante. Se usará solo FECHA_ACCIDENTE.')
```

## 6. Detección de formato inconsistente

```python
print('=== GRAVEDAD_ACCIDENTE ===')
print(df['GRAVEDAD_ACCIDENTE'].value_counts())
print('>>> PROBLEMA: Solo daños y Solo da\\xF1os son el mismo valor (encoding latin-1 vs utf-8).')
print('>>> Decisión ETL: unificar ambos como Solo daños.')

print('\n=== CLASE_ACCIDENTE ===')
print(df['CLASE_ACCIDENTE'].value_counts())
print('>>> PROBLEMA: Caida Ocupante, Caída Ocupante y Caída de Ocupante son la misma categoría.')
print('>>> Decisión ETL: normalizar todas a Caída de Ocupante.')

print('\n=== DISEÑO ===')
print(df['DISEÑO'].value_counts())
print('>>> PROBLEMA: Pontón y Pont\\xF3n son el mismo valor (encoding).')
print('>>> Decisión ETL: unificar como Pontón.')
```

## 7. Detección de outliers (coordenadas X, Y)

```python
fig = go.Figure()
fig.add_trace(go.Box(y=df['X'].dropna(), name='Coordenada X', marker_color='#1D9E75'))
fig.add_trace(go.Box(y=df['Y'].dropna(), name='Coordenada Y', marker_color='#378ADD'))
fig.update_layout(title='Distribución de coordenadas geográficas (X, Y)')
fig.show()

for col in ['X', 'Y']:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    outliers = df[(df[col] < Q1 - 1.5*IQR) | (df[col] > Q3 + 1.5*IQR)]
    print(f'Outliers en {col}: {len(outliers)} ({len(outliers)/len(df)*100:.2f}%)')

print('>>> Las coordenadas están dentro del rango de Medellín (MAGNA-SIRGAS). No se eliminan.')
```

## 8. Análisis exploratorio — distribuciones principales

```python
# Incidentes por año
df['AÑO_CLEAN'] = df['AÑO'].astype(str).str.strip().str.replace(r'\r', '', regex=True)
por_anio = df.groupby('AÑO_CLEAN').size().reset_index(name='total').sort_values('AÑO_CLEAN')

fig = px.bar(
    por_anio,
    x='AÑO_CLEAN', y='total',
    title='Incidentes viales por año',
    labels={'AÑO_CLEAN': 'Año', 'total': 'Total incidentes'},
    color='total',
    color_continuous_scale='Teal',
    text='total'
)
fig.update_traces(texttemplate='%{text:,}', textposition='outside')
fig.update_layout(showlegend=False)
fig.show()
```

```python
# Distribución por gravedad
gravedad = df['GRAVEDAD_ACCIDENTE'].value_counts().reset_index()
gravedad.columns = ['gravedad', 'total']

fig = px.pie(
    gravedad,
    names='gravedad',
    values='total',
    title='Distribución por gravedad del accidente',
    color_discrete_sequence=px.colors.qualitative.Safe
)
fig.update_traces(textposition='inside', textinfo='percent+label')
fig.show()
```

```python
# Top 10 comunas
top_comunas = (
    df[df['COMUNA'].notna()]
    .groupby('COMUNA').size()
    .nlargest(10)
    .reset_index(name='total')
    .sort_values('total')
)

fig = px.bar(
    top_comunas,
    x='total', y='COMUNA',
    orientation='h',
    title='Top 10 comunas con más incidentes viales',
    labels={'total': 'Total incidentes', 'COMUNA': 'Comuna'},
    color='total',
    color_continuous_scale='Purples',
    text='total'
)
fig.update_traces(texttemplate='%{text:,}', textposition='outside')
fig.update_layout(showlegend=False)
fig.show()
```

```python
# Clase de accidente
clase = df['CLASE_ACCIDENTE'].value_counts().reset_index()
clase.columns = ['clase', 'total']

fig = px.bar(
    clase,
    x='clase', y='total',
    title='Incidentes por clase de accidente',
    labels={'clase': 'Clase', 'total': 'Total'},
    color='total',
    color_continuous_scale='Blues',
    text='total'
)
fig.update_traces(texttemplate='%{text:,}', textposition='outside')
fig.update_layout(showlegend=False)
fig.show()
```

```python
# Estacionalidad mensual
por_mes = df.groupby('MES').size().reset_index(name='total')
meses = {1:'Ene',2:'Feb',3:'Mar',4:'Abr',5:'May',6:'Jun',
          7:'Jul',8:'Ago',9:'Sep',10:'Oct',11:'Nov',12:'Dic'}
por_mes['mes_nombre'] = por_mes['MES'].map(meses)

fig = px.line(
    por_mes,
    x='mes_nombre', y='total',
    title='Estacionalidad mensual de incidentes',
    labels={'mes_nombre': 'Mes', 'total': 'Total incidentes'},
    markers=True,
    color_discrete_sequence=['#D85A30']
)
fig.update_traces(line_width=2.5, marker_size=8)
fig.show()
```

```python
# Gravedad por año (apilado)
df['AÑO_CLEAN'] = df['AÑO'].astype(str).str.strip().str.replace(r'\r', '', regex=True)
por_anio_gravedad = df.groupby(['AÑO_CLEAN', 'GRAVEDAD_ACCIDENTE']).size().reset_index(name='total')

fig = px.bar(
    por_anio_gravedad,
    x='AÑO_CLEAN', y='total',
    color='GRAVEDAD_ACCIDENTE',
    title='Incidentes por año y gravedad',
    labels={'AÑO_CLEAN': 'Año', 'total': 'Total', 'GRAVEDAD_ACCIDENTE': 'Gravedad'},
    barmode='stack',
    color_discrete_sequence=px.colors.qualitative.Safe
)
fig.show()
```
